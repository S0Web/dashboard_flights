```python
# weather.pdf to weather.csv

HEADER = "origin,year,month,day,hour,temp,dewp,humid,wind_dir,wind_speed,wind_gust,precip,pressure,visib,time_hour"
lines = []
for page in PdfReader("weather.pdf").pages:
    for line in (page.extract_text() or "").splitlines():
        line = line.strip()
        if line and line != HEADER:
            lines.append(line)
bad = [l for l in lines if l.count(",") != 14]
if bad:
    print(f"Attention : {len(bad)} ligne(s) météo ignorée(s), exemple : {bad[0]}")
lines = [l for l in lines if l.count(",") == 14]
lines
weather = pd.DataFrame([l.split(",") for l in lines], columns=HEADER.split(","))
weather.to_csv("weather.csv", index=False)
# ("weather.csv").write_text(HEADER + "\n" + "\n".join(lines) + "\n", encoding="utf-8")
```
